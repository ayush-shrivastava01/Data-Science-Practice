```python
from abc import ABC, abstractmethod


# ---------------- ABSTRACT CLASS ----------------
class Coffee(ABC):

    def __init__(self, name, water, milk, coffee_powder):
        self.__name = name
        self.__water = water
        self.__milk = milk
        self.__coffee_powder = coffee_powder

    # Encapsulation using getter methods
    def get_name(self):
        return self.__name

    def get_water(self):
        return self.__water

    def get_milk(self):
        return self.__milk

    def get_coffee_powder(self):
        return self.__coffee_powder

    @abstractmethod
    def recipe(self):
        pass

    @abstractmethod
    def image(self):
        pass


# ---------------- CHILD CLASSES ----------------

class Latte(Coffee):

    def __init__(self):
        super().__init__("Latte", 100, 150, 2)

    def recipe(self):
        print("\nRecipe for Latte")
        print(f"Water : {self.get_water()} ml")
        print(f"Milk  : {self.get_milk()} ml")
        print(f"Coffee Powder : {self.get_coffee_powder()} spoons")

    def image(self):
        print("""
          ( (
           ) )
        ........
        | LATTE |
        |      |
        |______|
        """)


class Americano(Coffee):

    def __init__(self):
        super().__init__("Americano", 180, 20, 3)

    def recipe(self):
        print("\nRecipe for Americano")
        print(f"Water : {self.get_water()} ml")
        print(f"Milk  : {self.get_milk()} ml")
        print(f"Coffee Powder : {self.get_coffee_powder()} spoons")

    def image(self):
        print("""
          )))
         (((
        +------+
        |AMERI |
        |CANO  |
        +------+
        """)


class Matcha(Coffee):

    def __init__(self):
        super().__init__("Matcha", 120, 100, 1)

    def recipe(self):
        print("\nRecipe for Matcha")
        print(f"Water : {self.get_water()} ml")
        print(f"Milk  : {self.get_milk()} ml")
        print(f"Matcha Powder : {self.get_coffee_powder()} spoons")

    def image(self):
        print("""
         //////
        |MATCHA|
        |      |
        |______|
        """)


# ---------------- COFFEE MACHINE ----------------

class CoffeeMachine:

    def serve_coffee(self, coffee):
        print("\nPreparing your coffee...")
        coffee.image()
        coffee.recipe()
        print("\nYour coffee is ready. Enjoy!\n")


# ---------------- MAIN PROGRAM ----------------

machine = CoffeeMachine()

while True:

    print("====== COFFEE MACHINE ======")
    print("1. Latte")
    print("2. Americano")
    print("3. Matcha")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        drink = Latte()
        machine.serve_coffee(drink)

    elif choice == "2":
        drink = Americano()
        machine.serve_coffee(drink)

    elif choice == "3":
        drink = Matcha()
        machine.serve_coffee(drink)

    elif choice == "4":
        print("Coffee Machine Closed.")
        break

    else:
        print("Invalid Choice. Try Again.\n")
```
